```python
# ==============================================================================
# Step 1: Install Required Libraries
# ==============================================================================
!pip install fiftyone

# ==============================================================================
# Step 2: Import Libraries
# ==============================================================================
import fiftyone as fo
import fiftyone.zoo as foz
import os
import yaml

# ==============================================================================
# Step 3: Define Classes and Download Dataset
# ==============================================================================
open_images_classes = [
    "Person", "Bottle", "Chair", "Backpack", "Laptop",
    "Book", "Mobile phone", "Television", "Table", "Computer keyboard",
    "Car", "Couch", "Bench", "Waste container", "Dog"
]

print("Starting to download a balanced subset of the Open Images V7 dataset...")
print(f"The following classes will be downloaded: {open_images_classes}")

try:
    # Load training dataset
    train_dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="train",
        label_types=["detections"],
        classes=open_images_classes,
        max_samples=5000,
    )

    # Load validation dataset
    val_dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="validation",
        label_types=["detections"],
        classes=open_images_classes,
        max_samples=1000,
    )

    # Load test dataset
    test_dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="test",
        label_types=["detections"],
        classes=open_images_classes,
        max_samples=500,
    )

    # Rename 'ground_truth' to 'detections' for compatibility with YOLO export
    for dataset in [train_dataset, val_dataset, test_dataset]:
        if "ground_truth" in dataset.get_field_schema():
            dataset.rename_sample_field("ground_truth", "detections")

    # ==============================================================================
    # Step 4: Export to YOLO Format
    # ==============================================================================
    export_dir = "open-images-yolo"
    print(f"\nExporting datasets to YOLO format in directory: {export_dir}...")

    labels_field = "detections"

    # Export training set
    train_dataset.export(
        export_dir=os.path.join(export_dir, "train"),
        dataset_type=fo.types.YOLOv5Dataset,
        classes=open_images_classes,
        labels_field=labels_field,
    )

    # Export validation set
    val_dataset.export(
        export_dir=os.path.join(export_dir, "val"),
        dataset_type=fo.types.YOLOv5Dataset,
        classes=open_images_classes,
        labels_field=labels_field,
    )

    # Export test set
    test_dataset.export(
        export_dir=os.path.join(export_dir, "test"),
        dataset_type=fo.types.YOLOv5Dataset,
        classes=open_images_classes,
        labels_field=labels_field,
    )

    print("✅ Dataset download and YOLO export complete.")
    print(f"🗂️ YOLO dataset ready at: '{export_dir}'")

    # ==============================================================================
    # Step 5: Generate data.yaml for YOLOv5/YOLOv8
    # ==============================================================================
    data_yaml_content = {
        "path": os.path.abspath(export_dir),
        "train": "train/images",
        "val": "val/images",
        "test": "test/images",
        "nc": len(open_images_classes),
        "names": open_images_classes,
    }

    data_yaml_path = os.path.join(export_dir, "data.yaml")
    with open(data_yaml_path, 'w') as f:
        yaml.dump(data_yaml_content, f, default_flow_style=False)

    print(f"📄 Generated 'data.yaml' at: {data_yaml_path}")

except Exception as e:
    print(f"❌ An error occurred during dataset preparation: {e}")
```
